## Serialize

```python
s_2 = Settings()
s_2.input.ams.task = "SinglePoint"
s_2.input.forcefield
s_2.input.ams.properties.gradients = True
s_2.input.ams.many_2 = [1, 2, 3, 4]
s_2.input.ams.many = {1: "asd", 2: "sda", 3: 2312, 4: "sdasd"}
s_2.json_serialize()
```

```python
s_2 = Settings()
s_2.input.ams[(1, 2)] = 33
s_2.json_serialize()
```

```python
from pathlib import Path


path_file = Path("job_name.in")

s_2 = Settings()
s_2.input.ams.task = "SinglePoint"
s_2.input.forcefield
s_2.input.ams.properties.gradients = True
s_2.input.ams.many_2 = [1, 2, 3, 4]
s_2.input.ams.many = {1: "asd", 2: "sda", 3: 2312, 4: "sdasd"}
s_2.write_inp(path_file)
```

```python
s_3 = Settings.read_inp(path_file)
s_2 == s_3
```

```python
path_file.unlink(missing_ok=True)
```

Be aware that deserialization of keys become strings.

```python
s_2.compare(s_3)
```

```python
s_3.compare(s_2)
```

## Nested keys

```python
s = Settings()
s.set_nested("input.ams.task", "SinglePoint")
s.set_nested("input.ams.properties.gradients", True)
```

```python
s.print_settings()
```

```python
s.pop_nested("input.ams.task")
```

```python
s.print_settings()
```

```python
s.get_nested("input.ams.task", default=False)
```

```python
s.print_settings()
```

You can check the block presence with:

```python
bool(s.get_nested("input.ams", default=False)), bool(s.get_nested("input.ams.not_existent", default=False))
```

The settings will remain unchanged

```python
s.print_settings()
```

Note the usage of `simple_nesting` input.

```python
s = Settings()
s.set_nested("input.ams.task", "SinglePoint", simple_nesting=True)
s.print_settings()
```

```python
s = Settings()
s.set_nested("input.ams.task", "SinglePoint", simple_nesting=False)
s.print_settings()
```
