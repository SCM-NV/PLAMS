## Compare Many settings

```python
from scm.plams.tools.settings_analysis import compare_settings
from scm.plams.tools.table_formatter import format_in_table
from typing import Any, Dict, Tuple
```

```python
s = Settings()
s.input.ams.task = "SinglePoint"
s.input.dftb.energy = False
s.input.dftb.other = "yes"
s.input.dftb.list_values = [11, 22, 33, 44]
s.input.ams.properties.gradients = True

settings_list = []

for v in [1, 30, 200, 200, 5]:
    s_i = s.copy()
    s_i.set_nested("input.ams.moleculardynamics.temperature", v)
    if v == 30:
        s_i.set_nested("input.ams.moleculardynamics.special.run", True)
    settings_list.append(s_i)

for v in [333, 99, 99, 99, 333]:
    s_i = s.copy()
    s_i.set_nested("input.ams.moleculardynamics.frequency", v)
    settings_list.append(s_i)

for v in [[333, 11], [333, 11], [333, 11], [333, 500], [333, 500]]:
    s_i = s.copy()
    s_i.set_nested("input.ams.moleculardynamics.sweep_temps", v)
    settings_list.append(s_i)

len(settings_list)
```

```python
default_settings = Settings()
default_settings.input.ams.moleculardynamics.frequency = 99
default_settings.input.ams.moleculardynamics.temperature = None
comparison_sett: Dict[Tuple, Any] = compare_settings(
    settings_list,
    analyze_blocks=False,
    analyze_keys=True,
    default_settings=default_settings,
    flatten_list=True,
    remove_unimportant_columns=True,
    none_is_unimportant=False,
)
comparison_sett_2 = {".".join(map(str, k)): v for k, v in comparison_sett.items()}
format_in_table(comparison_sett_2)
```

```python
s = Settings()
s.input.ams.task = "SinglePoint"
s.input.ams.properties.gradients = True

settings_list = []

for v in [1, 30, 200, 200, 5]:
    s_i = s.copy()
    s_i.set_nested("input.dftb.energy", v)
    settings_list.append(s_i)

for v in [333, 99, 99, 99, 333]:
    s_i = s.copy()
    s_i.set_nested("input.forcefield.charges", v)
    settings_list.append(s_i)


len(settings_list)
```

```python
default_settings = Settings()
default_settings.input.ams.moleculardynamics.frequency = 99
default_settings.input.ams.moleculardynamics.temperature = None
comparison_sett = compare_settings(
    settings_list,
    analyze_keys=True,
    default_settings=default_settings,
    flatten_list=True,
    remove_unimportant_columns=True,
)
comparison_sett_2 = {".".join(map(str, k)): v for k, v in comparison_sett.items()}
format_in_table(comparison_sett_2)
```

## FreeBlocks conversion

```python
ml_s = Settings()
ml_s.input.MachineLearning.Backend = "Custom"
ml_s.input.MachineLearning.CommitteeSize = "2"
ml_s.input.MachineLearning.Custom.Arguments._1 = [
    "response_properties_and_loss = {}",
    "hparams_architecture = {'n_atom_basis': 50, 'n_rbf': 32, 'n_interactions': 2, 'cutoff': 5}",
    "hparams_training = {'optimizer': 1, 'lr_initial': 0.001, 'patience_lr_scheduler': 50, 'batch_size': 10}",
]
ml_s.input.MachineLearning.Custom.File = (
    "/home/benedini/branches_ams/scripting/scm/params/machine_learning/schnet/job.py"
)
ml_s.input.MachineLearning.LossCoeffs.Energy = "5.0"
ml_s.input.MachineLearning.LossCoeffs.Forces = "4.0"
ml_s.input.MachineLearning.MaxEpochs = "100"
ml_s.input.MachineLearning.RunAMSAtEnd = "True"
ml_s.input.ParallelLevels.CommitteeMembers = "2"
ml_s.input.Task = "MachineLearning"
ml_s.print_settings()
```

```python
ml_s_converted = ml_s.convert_free_block("input.MachineLearning.Custom.Arguments._1", inplace=False)
```

```python
assert ml_s_converted is not None
ml_s_converted.print_settings()
```
